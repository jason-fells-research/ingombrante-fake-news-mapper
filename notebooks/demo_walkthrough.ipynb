## 📚 Key Learning Points

Based on your analysis, here are the main patterns to watch for:

### 🔴 Red Flags (Suspicious Indicators):
- **Sensational Headlines**: Excessive caps, multiple exclamation marks, "SHOCKING" language
- **Vague Sources**: "Anonymous experts", "undisclosed study", "sources say"
- **Emotional Language**: Designed to provoke rather than inform
- **Suspicious Domains**: Non-established news sources, unusual URLs
- **Unverifiable Claims**: "Secret research", "suppressed information"

### 🟢 Trust Indicators (Credible Signs):
- **Named Sources**: Specific experts, institutions, studies with citations
- **Balanced Tone**: Informative rather than emotional
- **Established Sources**: Recognized news organizations
- **Verifiable Facts**: Specific dates, locations, statistics
- **Multiple Confirmations**: Cross-referenced information

### 💡 Pro Tips:
1. **Check the source** - Is it a known, reputable organization?
2. **Look for citations** - Are claims backed by named experts or studies?
3. **Examine the tone** - Is it trying to inform or provoke?
4. **Cross-reference** - Can you find the same story from multiple reliable sources?
5. **Be skeptical of perfection** - Real news often includes nuance and uncertainty
